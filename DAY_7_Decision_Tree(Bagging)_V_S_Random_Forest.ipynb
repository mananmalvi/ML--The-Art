```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
```

```python
x,y=make_classification(n_samples=1000, n_features=5, n_redundant=0, n_informative=2, random_state=42, n_clusters_per_class=1)
```

```python
import pandas as pd
# Define all 6 columns, explicitly naming the last one as 'target'
df = pd.DataFrame(x, columns=['cols1', 'col2', 'col3', 'col4', 'target'])
# Now this will extract successfully!
y = df['target']

print(df.shape)
df.head()
```

FOR BAGGING

```python
from sklearn.tree import plot_tree
from sklearn.ensemble import BaggingClassifier
from sklearn.preprocessing import LabelEncoder
# by default base estimator is decision tree
bag = BaggingClassifier(n_estimators=100, max_samples =1.0, n_jobs=-1, max_features=2)
X = df.iloc[:, :4]
y_raw = df.iloc[:, -1]
y_encoded = LabelEncoder().fit_transform(y_raw)
bag.fit(X, y_encoded)
```

```python
# plotting
from sklearn.tree import plot_tree
plt.figure(figsize=(12,12))
plot_tree(bag.estimators_[0])
plt.show()
```

FOR RANDOM FOREST

```python
from sklearn.ensemble import RandomForestClassifier
rf= RandomForestClassifier(n_estimators=100, n_jobs=-1)
rf.fit(X,y_encoded)
```
