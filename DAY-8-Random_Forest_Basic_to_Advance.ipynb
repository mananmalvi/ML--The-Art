# Random Forest -Basic to Advance

# IMPORTING DATA

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/content/heart.csv')
df.head()
df.shape
```

```python
x=df.iloc[:,0:-1]
y=df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

# MODEL-1-🌲 MOST BASIC RANDOM FOREST MODEL 🤖🌳

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score
rf=RandomForestClassifier()
rf.fit(x_train,y_train)
y_pred=rf.predict(x_test)

print(f"Test Accuracy: {(accuracy_score(y_test, y_pred))*100}")
```

# MODEL-2- 🤖🌳 MODEL 2 — USING MULTIPLE MODELS 🔥🧠

```python
from sklearn.svm import SVC
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Create models
svc = SVC()
gbc = GradientBoostingClassifier()
lrc = LogisticRegression()

# Train models
svc.fit(x_train, y_train)
gbc.fit(x_train, y_train)
lrc.fit(x_train, y_train)

# Predictions
y_pred_svc = svc.predict(x_test)
y_pred_gbc = gbc.predict(x_test)
y_pred_lrc = lrc.predict(x_test)

# Accuracy
print(f"Test Accuracy SVC: {accuracy_score(y_test, y_pred_svc) * 100:.2f}%")
print(f"Test Accuracy GBC: {accuracy_score(y_test, y_pred_gbc) * 100:.2f}%")
print(f"Test Accuracy LRC: {accuracy_score(y_test, y_pred_lrc) * 100:.2f}%")
```

# MODEL-3- 🤖🌳 MODEL 3 — RANDOM FOREST CLASSIFIER (USING HYPERPARAMETER) 🔥🧠

```python
from sklearn.ensemble import RandomForestClassifier
rf_hyper=RandomForestClassifier(
    max_samples=0.75,
    max_features="sqrt",
    n_estimators=100,
    criterion="gini",
    bootstrap=True,
    max_depth=10,
    min_samples_split=2,
    oob_score=True,
    random_state=42
)

rf_hyper.fit(x_train,y_train)

y_pred_rf_hyper=rf_hyper.predict(x_test)

print(f"Test Accuracy: {(accuracy_score(y_test, y_pred_rf_hyper))*100}")
```

# MODEL-4- 🤖🌳 MODEL 4 — RANDOM FOREST CLASSIFIER (USING GRID SEARCH CV) 🔥🧠

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

# Ensure x_train and y_train are defined and preprocessed before execution.

param_grid = {
    'n_estimators': [50, 100, 250, 500],
    'criterion': ['gini', 'entropy', 'log_loss'],
    'max_depth': [None, 10, 20, 50],
    'min_samples_split': [2, 5, 10],
    'max_features': ['sqrt', 'log2', None],
    'max_leaf_nodes': [None, 50, 100],
    'bootstrap': [True], # Forced True to allow max_samples compatibility
    'max_samples': [None, 0.4, 0.7, 0.9]
}

# Initialize model
rf = RandomForestClassifier(random_state=42, n_jobs=-1)

# Initialize grid search
# WARNING: This exhaustive grid is computationally massive.
rf_grid = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=5,
    n_jobs=-1,
    verbose=2,
    scoring='accuracy' # Specify the metric to optimize
)

# Execute optimization
rf_grid.fit(x_train, y_train)

# Output extracted high-fidelity intelligence
print(f"Optimal Parameters: {rf_grid.best_params_}")
print(f"Peak CV Score: {rf_grid.best_score_}")
```

# MODEL-5- 🤖🌳 MODEL 5 — RANDOM FOREST CLASSIFIER (USING RANDOMIZED SEARCH CV) 🔥🧠

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier

rf=RandomForestClassifier(random_state=42, n_jobs=-1)

# Define the parameter grid

param_grid = {
    'n_estimators': [50, 100, 250, 500],
    'criterion': ['gini', 'entropy', 'log_loss'],
    'max_depth': [None, 10, 20, 50],
    'min_samples_split': [2, 5, 10],
    'max_features': ['sqrt', 'log2', None],
    'max_leaf_nodes': [None, 50, 100],
    'bootstrap': [True], # Forced True to allow max_samples compatibility
    'max_samples': [None, 0.4, 0.7, 0.9]
}

rf_random_grid=RandomizedSearchCV(
    estimator=rf,
    param_distributions=param_grid,
    n_iter=100,
    cv=5,
    n_jobs=-1,
    verbose=2,
    scoring='accuracy'
)

rf_random_grid.fit(x_train,y_train)

print(f"Optimal Parameters: {rf_random_grid.best_params_}")
print(f"Peak CV Score: {rf_random_grid.best_score_}")
```
