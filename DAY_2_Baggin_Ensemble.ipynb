2. BASIC BAGGING CLASSIFICATION MODEL

3.BASIC BAGGING REGRESSION MODEL

```python
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import BaggingRegressor
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
```

```python
x,y =make_regression(n_samples=1000, n_features=10, n_informative=10,random_state=42 )
```

```python
x_train, x_test, y_train,y_test = train_test_split(x,y,test_size=0.2, random_state=42)
```

```python
from sklearn.tree import DecisionTreeRegressor
dt=DecisionTreeRegressor(random_state=42)
```

```python
model2= BaggingRegressor(estimator=dt, n_estimators=10,max_samples=0.8,max_features=1.0, bootstrap=True, random_state=42,n_jobs=-1)
```

```python
model2.fit(x_train, y_train)
```

```python
y_pred=model2.predict(x_test)
```

```python
print('Mean Squared Error:', mean_squared_error(y_test, y_pred))
print('R-squared:', r2_score(y_test, y_pred)*100)
print('Mean Absolute Error:', mean_absolute_error(y_test, y_pred))
```

TYPES OF BAGGING -
1.Bagging — Samples + Replacement
2.Pasting-Random subsets of samples
3.Random Subspaces-Yahan samples nahi, features random select hote hain
4.Random Patches-Dono random:

 1. Bagging — Samples + Replacement

```python
from sklearn.ensemble import BaggingRegressor
base_model=DecisionTreeRegressor(random_state=42)

model3 = BaggingRegressor(
    estimator=base_model,
    n_estimators=100,

    max_samples=0.8,      # 80% samples
    bootstrap=True,       # ✅ With replacement

    max_features=1.0,     # Use all features
    bootstrap_features=False,

    random_state=42,
    n_jobs=-1)
```

2. Pasting (Random Samples WITHOUT Replacement)

```python
from sklearn.ensemble import BaggingRegressor

model4 = BaggingRegressor(
    estimator=base_model,
    n_estimators=100,

    max_samples=0.8,
    bootstrap=False,      # ❌ No replacement

    max_features=1.0,
    bootstrap_features=False,

    random_state=42,
    n_jobs=-1
)
```

3. Random Subspaces (Random Features)

```python
from sklearn.ensemble import BaggingRegressor
# Definition: Random subsets of features are selected

model5 = BaggingRegressor(
    estimator=base_model,
    n_estimators=100,

    max_samples=1.0,      # Use all samples
    bootstrap=True,

    max_features=0.6,     # 60% features
    bootstrap_features=False,

    random_state=42,
    n_jobs=-1
)
```

4. Random Patches (Random Samples + Random Features)

```python
# Definition: Random subsets of both samples and features are selected.
from sklearn.ensemble import BaggingRegressor

model = BaggingRegressor(
    estimator=base_model,
    n_estimators=100,

    max_samples=0.8,      # Random samples
    bootstrap=True,

    max_features=0.6,     # Random features
    bootstrap_features=False,

    random_state=42,
    n_jobs=-1
)
```

Bagging Tips
Bagging generally gives better results than Pasting
Good results come around the 25% to 50% row sampling mark
Random patches and subspaces should be used while dealing with high dimensional data
To find the correct hyperparameter values we can do GridSearchCV/RandomSearchCV
